# AD Clinical Trial Landscape — Quick Analysis
Run `adtrial collect` first.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
ROOT=Path('..').resolve()
studies=pd.read_csv(ROOT/'data/processed/studies.csv',keep_default_na=False)
interventions=pd.read_csv(ROOT/'data/processed/interventions.csv',keep_default_na=False)
studies.head()
```

```python
ps=studies.groupby(['phase','overall_status']).size().reset_index(name='Trials')
px.bar(ps,x='phase',y='Trials',color='overall_status',barmode='stack',title='Phase × Status')
```

```python
mech=interventions['mechanism_category'].value_counts().rename_axis('Mechanism').reset_index(name='Interventions')
px.bar(mech.sort_values('Interventions'),x='Interventions',y='Mechanism',orientation='h',title='Mechanism Landscape')
```
